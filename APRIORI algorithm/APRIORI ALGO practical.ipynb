```python
import pandas as pd

pd.set_option('display.max_rows', 200)
dataset=pd.read_csv('groceries - groceries.csv')
dataset.head()
```

```python
now, we wil search the highest selleing products and combined highest selling products via support
and then via confidence the arrangement of products can be determined
```

# first of all remove NaN values

```python
type(dataset['Item 6'][0]) # float
type(dataset['Item 1'][0]) # str

# yani NaN ka type float h
```

```python
market=[]
for row in range(0,dataset.shape[0]):
    cus=[]
    for col in dataset.columns:
        if(type(dataset[col][row])==str):
            cus.append(dataset[col][row])
    market.append(cus)
market
```

```python
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
te=TransactionEncoder()
te.fit(market) 
```

```python
mat=te.transform(market)
mat
#yani hmy ek matrix chahye jis m saaaarrrryy products k name as columns hon and 
# phir btaya ja rha ho k is customer n in products m s y y buy kia h , jha jha true aya yani us transaction m y products
# purchase hui hn and jha jha false aya yani buy nhi ki gi
```

```python
matrix=pd.DataFrame(mat, columns=te.columns_)
matrix.head() 
```

```python

from mlxtend.frequent_patterns import apriori
d = apriori(matrix,min_support=0.05,use_colnames=True, max_len=3)  
# agr max_length=None rkhi to vo saary products show hon gy jin ki support 5% s ziada thi, ab 
# chahy vo koi ek ek products show hon ya phir 2 2 k combination ya 3 3 k...
# but agr hm n length fix kr di yani max_length=3 rkhi to vo products jin ki support 5% s ziada
# thi vo products hi show hon gy but max_len = 3 rkh kr hm n y bta di k vo products show hon jin
# ki support 5% s ziada ho and un k combination 3 hi hon, ab vo
# combination 1 bhi ho skta (yani 1 1 products bhi show ho skti jin ki support 5% s ziada h), 2 
# bhi and 3 bhi, but 3 s ziada nhi ho skta,,, agr hm n combination 3 ka kha tha and 3 nhi show 
# ho rhi (bal k just 1 and 2 hi show ho rhi) to esa is
# liye hua k combination of 3 products is criterion pr poor hi nhi hua k un ki min_support 5% ho

# -----> ab min-support ki value 0.05 is liye rkhi q k y bhhttt bra dataset h, poory store ka 
# and itny bry store m possible nhi hota k ek saath bht saari products ek saath bhhhttt log
# buy kryn akathy, and support ki value 0.05 pr bhi 2 products
# jo ek sath buy hui vo km hn
d
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(
    d,
    metric="confidence",
    min_threshold=0.1
)  #there are 100s of rukles for filtering, hm bs y keh rhy hn k vo vo products show krny hn jin ka 
# confidence minimum 0.1 ho, yani jin ka confidence 0.1 k equal ya us s greater hi bs vo rkho
# baki sab ko nikal do 

rules
print(rules[['antecedents',
             'consequents',
             'support',
             'confidence',
             'lift']])
```
